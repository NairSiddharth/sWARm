**Objectives:**
- Use K-fold CV to avoid overfitting

**Key Features:**
- GroupKFold to keep years together (prevents data leakage)
- 4 subplots per year: Hitter WAR, Hitter WARP, Pitcher WAR, Pitcher WARP

## Individual Year Analysis

You can also analyze specific years individually:

```python
# Example: Analyze a specific year in detail
target_year = '2021'  # Change this to analyze different years

print(f"Detailed analysis for {target_year}:")
print("-" * 40)

# Create detailed graph for this year
fig_detail = plot_year_specific_analysis(cv_results, target_year)

if fig_detail:
    # Show the graph
    fig_detail.show()
    
    # Print statistics for this year
    year_data = cv_results.get_year_data(target_year)
    print(f"\nStatistics for {target_year}:")
    
    for key, data in year_data.items():
        if len(data['y_true']) > 0:
            from sklearn.metrics import r2_score, mean_squared_error
            r2 = r2_score(data['y_true'], data['y_pred'])
            rmse = np.sqrt(mean_squared_error(data['y_true'], data['y_pred']))
            
            model_name, player_type, metric_type = key.split('_')
            print(f"  {model_name.title()} {player_type} {metric_type.upper()}: "
                  f"{len(data['y_true'])} predictions, R²={r2:.4f}, RMSE={rmse:.4f}")
else:
    print(f"No data available for {target_year}")
```

## Validation and Testing

```python
print("STEP 6: Validation and testing")
print("="*70)

# Validate that every player was predicted exactly once
total_predictions = 0
unique_players = set()

for key, data in cv_results.results.items():
    total_predictions += len(data['y_true'])
    for player, year in zip(data['player_names'], data['years']):
        unique_players.add(f"{player}_{year}")

print(f"Total predictions made: {total_predictions}")
print(f"Unique player-year combinations: {len(unique_players)}")

# Check for year distribution
year_counts = {}
for key, data in cv_results.results.items():
    for year in data['years']:
        year_str = str(year)
        year_counts[year_str] = year_counts.get(year_str, 0) + 1

print(f"\nPredictions per year:")
for year in sorted(year_counts.keys()):
    print(f"  {year}: {year_counts[year]} predictions")

# Verify no data leakage (each player predicted exactly once per model/category)
for key, data in cv_results.results.items():
    player_years = [f"{p}_{y}" for p, y in zip(data['player_names'], data['years'])]
    if len(player_years) != len(set(player_years)):
        print(f"WARNING: Duplicate predictions found in {key}")
    else:
        print(f"VERIFIED: No duplicates in {key} ({len(player_years)} unique predictions)")

print("\nValidation complete!")
```
