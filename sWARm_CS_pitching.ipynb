# Step 2: Train Ensemble Models on Historical Data

```python
print("STEP 2: Training Ensemble Models on Historical Data (2016-2024)")
print("=" * 60)
print("PHASE 1 INTEGRATION: Rate-based three-path ensemble for pitchers")
print("=" * 60)

# Import ensemble modeling functions
from current_season_modules.modeling.ensemble_modeling import create_ensemble_for_data
from current_season_modules.modeling.pitcher_roles_ensemble_standalone import train_three_path_ensemble

import pandas as pd
import numpy as np
import glob

if hitter_data_dict or pitcher_data_dict:
    
    # ========================================
    # PART 1: Train Phase 1 Ensemble for Pitchers
    # ========================================
    
    if pitcher_data_dict:
        print("\n[PITCHERS] Training Phase 1 rate-based three-path ensemble...")
        print("-" * 60)
        
        # Load G and GS from raw FanGraphs CSV files
        print("Loading G and GS from raw FanGraphs CSV files...")
        gs_g_data = {}  # key: (year, pitcher_name) -> (G, GS)
        
        csv_pattern = "MLB Player Data/FanGraphs_Data/pitchers/fangraphs_pitchers_*_standard.csv"
        csv_files = glob.glob(csv_pattern)
        
        for csv_file in csv_files:
            try:
                df = pd.read_csv(csv_file)
                # Extract year from filename (skip firsthalf files)
                filename_parts = csv_file.split('_')
                year_str = filename_parts[-2]
                try:
                    year = int(year_str)
                except ValueError:
                    continue  # Skip non-year files like firsthalf
                
                if 'G' in df.columns and 'GS' in df.columns and 'Name' in df.columns:
                    for _, row in df.iterrows():
                        name = row['Name']
                        gs_g_data[(year, name)] = (row['G'], row['GS'])
            except Exception as e:
                print(f"Warning: Could not load {csv_file}: {e}")
                continue
        
        print(f"Loaded G/GS data for {len(gs_g_data)} pitcher-seasons")
        
        # Convert pitcher_data_dict to Phase 1 format (DataFrames with G/GS)
        pitcher_dfs = {}
        for metric in ['war', 'warp']:
            if metric in pitcher_data_dict:
                data = pitcher_data_dict[metric]
                
                # Extract components
                X_df = data['X']  # Already a DataFrame
                y_series = data['y']
                years_data = data['years']
                names_data = data['names']
                
                n_samples = len(X_df)
                
                # Convert years to flat list - handle different structures
                if isinstance(years_data, (list, tuple)) and len(years_data) > 0:
                    # Check if it's nested (list of lists)
                    if isinstance(years_data[0], (list, tuple)):
                        # Flatten nested structure
                        years_list = [item[0] if isinstance(item, (list, tuple)) else item for item in years_data]
                    else:
                        years_list = list(years_data)
                elif hasattr(years_data, 'values'):
                    # Pandas Series
                    years_array = years_data.values
                    if len(years_array) > 0 and isinstance(years_array[0], (list, tuple)):
                        years_list = [item[0] if isinstance(item, (list, tuple)) else item for item in years_array]
                    else:
                        years_list = list(years_array)
                elif isinstance(years_data, np.ndarray):
                    if years_data.dtype == object and len(years_data) > 0 and isinstance(years_data[0], (list, tuple)):
                        years_list = [item[0] if isinstance(item, (list, tuple)) else item for item in years_data]
                    else:
                        years_list = list(years_data)
                else:
                    years_list = [2024] * n_samples
                
                # Convert names to flat list - handle different structures
                if isinstance(names_data, (list, tuple)) and len(names_data) > 0:
                    if isinstance(names_data[0], (list, tuple)):
                        names_list = [item[0] if isinstance(item, (list, tuple)) else item for item in names_data]
                    else:
                        names_list = list(names_data)
                elif hasattr(names_data, 'values'):
                    names_array = names_data.values
                    if len(names_array) > 0 and isinstance(names_array[0], (list, tuple)):
                        names_list = [item[0] if isinstance(item, (list, tuple)) else item for item in names_array]
                    else:
                        names_list = list(names_array)
                elif isinstance(names_data, np.ndarray):
                    if names_data.dtype == object and len(names_data) > 0 and isinstance(names_data[0], (list, tuple)):
                        names_list = [item[0] if isinstance(item, (list, tuple)) else item for item in names_data]
                    else:
                        names_list = list(names_data)
                else:
                    names_list = ["Unknown"] * n_samples
                
                # Ensure lists match sample count
                if len(years_list) != n_samples:
                    print(f"  WARNING {metric}: years length {len(years_list)} != samples {n_samples}, using defaults")
                    years_list = [2024] * n_samples
                if len(names_list) != n_samples:
                    print(f"  WARNING {metric}: names length {len(names_list)} != samples {n_samples}, using defaults")
                    names_list = ["Unknown"] * n_samples
                
                # Get G and GS arrays from CSV data
                G_array = np.zeros(n_samples)
                GS_array = np.zeros(n_samples)
                
                for i in range(n_samples):
                    year = int(years_list[i])
                    name = str(names_list[i])
                    key = (year, name)
                    
                    if key in gs_g_data:
                        G_array[i], GS_array[i] = gs_g_data[key]
                    else:
                        # Fallback defaults (reliever-like)
                        G_array[i] = 20
                        GS_array[i] = 0
                
                matched = np.sum(G_array > 0)
                print(f"  {metric.upper()}: Matched G/GS for {matched}/{n_samples} pitchers")
                
                # Create DataFrame with all needed columns
                pitcher_df = X_df.copy()
                pitcher_df['Season'] = years_list
                pitcher_df['G'] = G_array
                pitcher_df['GS'] = GS_array
                pitcher_df[metric.upper()] = y_series.values if hasattr(y_series, 'values') else y_series
                
                pitcher_dfs[metric] = pitcher_df
        
        # Train Phase 1 ensemble (no holdout for production)
        pitcher_ensemble = train_three_path_ensemble(
            pitcher_dfs,
            holdout_year=None  # Use all data for production
        )
        
        print("✓ Phase 1 pitcher ensemble trained (rate-based, three-path routing)")
    else:
        pitcher_ensemble = None
        print("⚠ No pitcher data available")
    
    # ========================================
    # PART 2: Train Original Ensemble for Hitters
    # ========================================
    
    if hitter_data_dict:
        print("\n[HITTERS] Training original ensemble (volume-based)...")
        print("-" * 60)
        
        hitter_ensemble = create_ensemble_for_data(
            hitter_data_dict,
            {},  # No pitchers - handled by Phase 1
            holdout_year=None  # Use all data for production
        )
        
        print("✓ Hitter ensemble trained (volume-based, RF+Keras)")
    else:
        hitter_ensemble = None
        print("⚠ No hitter data available")
    
    # ========================================
    # SUMMARY
    # ========================================
    
    print("\n" + "=" * 60)
    print("ENSEMBLE TRAINING COMPLETE")
    print("=" * 60)
    
    if hitter_ensemble:
        print("✓ Hitter ensemble ready (original method)")
        validation_summary = hitter_ensemble.get_validation_summary()
        for key, results in validation_summary.items():
            if 'hitter' in key:
                print(f"  - {results['metric_type'].upper()}: R² = {results['ensemble_performance']:.4f}")
    
    if pitcher_ensemble:
        print("✓ Pitcher ensemble ready (Phase 1: rate-based, three-path)")
        print("  - WAR: Trained on rate stats (WAR per 162 IP)")
        print("  - WARP: Trained on rate stats (WARP per 162 IP)")
        print("  - Three paths: Starters, Relievers, Mixed roles")
    
    print("\nEnsemble models ready for 2025 projections!")
    
    # For backward compatibility with later cells, create a wrapper
    # that dispatches to the appropriate ensemble
    class EnsembleWrapper:
        def __init__(self, hitter_ens, pitcher_ens):
            self.hitter_ensemble = hitter_ens
            self.pitcher_ensemble = pitcher_ens
        
        def predict_ensemble(self, features, metric_type, player_type):
            """Backward-compatible prediction interface."""
            if player_type == 'hitter' and self.hitter_ensemble:
                return self.hitter_ensemble.predict_ensemble(features, metric_type, player_type)
            elif player_type == 'pitcher' and self.pitcher_ensemble:
                # Phase 1 requires different interface - this is just for compatibility
                # Real pitcher predictions should use pitcher_ensemble.predict() directly
                # For now, return a dummy result - will be overridden in Cell 12
                return {'ensemble': 0.0}
            else:
                return {'ensemble': 0.0}
        
        def get_validation_summary(self):
            """Get validation summary from hitter ensemble."""
            if self.hitter_ensemble:
                return self.hitter_ensemble.get_validation_summary()
            return {}
    
    ensemble_predictor = EnsembleWrapper(hitter_ensemble, pitcher_ensemble)
    
else:
    print("⚠ No historical data available for model training")
    print("Cannot proceed with projections without trained models")
    ensemble_predictor = None
    hitter_ensemble = None
    pitcher_ensemble = None
```

```python
import pandas as pd
import numpy as np

print("STEP 3: Loading First Half 2025 Data (ALL CSVs + Fresh Composites)")
print("=" * 70)

# Load ALL three CSV files for complete feature set
csv_standard_path = "MLB Player Data/FanGraphs_Data/pitchers/fangraphs_pitchers_2025_firsthalf_standard.csv"
csv_battedball_path = "MLB Player Data/FanGraphs_Data/pitchers/fangraphs_pitchers_2025_firsthalf_battedball.csv"
csv_advanced_path = "MLB Player Data/FanGraphs_Data/pitchers/fangraphs_pitchers_2025_firsthalf_advanced.csv"

print("Loading all 2025 first half CSV files...")

# Load hitters (unchanged - works correctly)
csv_hitters_path = "MLB Player Data/FanGraphs_Data/hitters/fangraphs_hitters_2025_firsthalf.csv"
first_half_hitters_raw = None
try:
    if os.path.exists(csv_hitters_path):
        first_half_hitters_raw = pd.read_csv(csv_hitters_path)
        print(f"[OK] Loaded hitters from CSV: {len(first_half_hitters_raw)} players")
    else:
        print(f"[WARNING] CSV not found: {csv_hitters_path}")
except Exception as e:
    print(f"[ERROR] Error loading hitters: {e}")

# Load ALL pitcher CSVs and merge
first_half_pitchers_raw = None
try:
    df_standard = pd.read_csv(csv_standard_path)
    print(f"[OK] Loaded standard CSV: {len(df_standard)} pitchers")

    df_battedball = pd.read_csv(csv_battedball_path)
    print(f"[OK] Loaded battedball CSV: {len(df_battedball)} pitchers")

    df_advanced = pd.read_csv(csv_advanced_path)
    print(f"[OK] Loaded advanced CSV: {len(df_advanced)} pitchers")

    # Merge on MLBAMID (or mlbid if MLBAMID not present)
    id_col = 'MLBAMID' if 'MLBAMID' in df_standard.columns else 'mlbid'

    # Select only needed columns from battedball and advanced to avoid duplicates
    battedball_cols = [id_col, 'Hard%', 'Med%', 'Soft%', 'HR/FB', 'LD%', 'GB%', 'FB%']
    battedball_cols = [col for col in battedball_cols if col in df_battedball.columns]

    advanced_cols = [id_col, 'K%', 'BB%', 'K-BB%']
    advanced_cols = [col for col in advanced_cols if col in df_advanced.columns]

    # Merge all three DataFrames
    first_half_pitchers_raw = df_standard.merge(
        df_battedball[battedball_cols], on=id_col, how='left'
    ).merge(
        df_advanced[advanced_cols], on=id_col, how='left'
    )

    print(f"[OK] Merged all CSVs: {len(first_half_pitchers_raw)} pitchers with {len(first_half_pitchers_raw.columns)} columns")

except Exception as e:
    print(f"[ERROR] Error loading/merging pitcher CSVs: {e}")
    import traceback
    traceback.print_exc()

# Process hitters (unchanged - works correctly)
print(f"\nSTEP 3B: Processing Hitters")
print("-" * 70)

from legacy_modules.historical_feature_preparation import prepare_historical_compatible_data

try:
    prepared_data = prepare_historical_compatible_data(first_half_hitters_raw, None)
    first_half_hitters = prepared_data['hitters']

    if first_half_hitters:
        print(f"\nProcessed First Half 2025 Hitters:")
        print(f"  Valid players: {len(first_half_hitters['valid_players'])}")
        print(f"  Feature matrix shape: {first_half_hitters['feature_matrix'].shape}")
        print(f"  Features: 10 [K%, BB%, AVG, OBP, SLG, PA, Position_Adj, GDP_rate, Enhanced_Baserunning, Enhanced_Defense]")

except Exception as e:
    print(f"[ERROR] Error processing hitters: {e}")
    first_half_hitters = None
    import traceback
    traceback.print_exc()

# Process pitchers with FRESH composite calculation + FULL park factors + PERCENTAGE FIX
print(f"\nSTEP 3C: Processing Pitchers (FRESH COMPOSITES + FULL PARK FACTORS + PERCENTAGE FIX)")
print("-" * 70)

from common_modules.park_factors import apply_park_factor_adjustments

def calculate_damage_control_ratio(hr_fb, hard_pct, soft_pct, med_pct):
    """Calculate damage control ratio from batted ball metrics."""
    # HR/FB weight: 40%, Hard% weight: 30%, Soft% inverse weight: 30%
    hr_fb_norm = min(hr_fb * 100, 100)
    hard_weight = hard_pct * 0.6
    soft_weight = (100 - soft_pct) * 0.4
    ratio = (hr_fb_norm * 0.4 + hard_weight * 0.3 + soft_weight * 0.3)
    return np.clip(ratio, 20, 80)

def calculate_contact_quality_index(hard_pct, med_pct, soft_pct, ld_pct, gb_pct, fb_pct):
    """Calculate contact quality index from batted ball distribution."""
    # Higher Hard% and LD% = worse (higher index)
    # Higher Soft% and GB% = better (lower index)
    hard_component = hard_pct * 0.4
    ld_component = ld_pct * 0.3
    soft_component = (100 - soft_pct) * 0.2
    gb_component = (100 - gb_pct) * 0.1

    index = hard_component + ld_component + soft_component + gb_component
    return np.clip(index, 30, 70)

def calculate_statcast_launch_quality(hr_fb, hard_pct):
    """Calculate launch quality index (proxy using HR/FB and Hard%)."""
    # Higher HR/FB and Hard% = worse (higher index)
    hr_fb_component = min(hr_fb * 100, 100) * 0.6
    hard_component = hard_pct * 0.4
    index = hr_fb_component + hard_component
    return np.clip(index, 30, 70)

def extract_pitcher_features_2025_fresh(pitcher_df):
    """Extract 10 pitcher features with FRESH composite calculation + FULL park factors + PERCENTAGE FIX."""

    feature_matrix = []
    valid_players = []
    player_names = []
    park_factors_applied = 0
    composites_calculated = 0

    for idx, row in pitcher_df.iterrows():
        name = row.get('Name', '')
        player_id = row.get('MLBAMID', row.get('mlbid', ''))
        team = row.get('Team', '')

        # Extract base stats
        ip = row.get('IP', 0)
        era = row.get('ERA', 0)
        bb_pct = row.get('BB%', 0)
        k_pct = row.get('K%', 0)
        k_bb_pct = row.get('K-BB%', 0)
        hbp_pct = row.get('HBP%', 0)

        # Batted ball stats - CONVERT DECIMALS TO PERCENTAGES (CRITICAL FIX)
        hr_fb = row.get('HR/FB', 0)
        if pd.notna(hr_fb):
            # If stored as decimal (0.085), convert to percentage (8.5)
            # If already percentage (8.5), keep as is
            hr_fb = hr_fb * 100 if hr_fb < 1 else hr_fb
            hr_fb = hr_fb / 100  # Then convert to decimal for calculation
        else:
            hr_fb = 0

        hard_pct = row.get('Hard%', 0)
        if pd.notna(hard_pct) and hard_pct < 1:
            hard_pct = hard_pct * 100  # Convert 0.263 -> 26.3
        elif pd.isna(hard_pct):
            hard_pct = 0

        med_pct = row.get('Med%', 0)
        if pd.notna(med_pct) and med_pct < 1:
            med_pct = med_pct * 100
        elif pd.isna(med_pct):
            med_pct = 0

        soft_pct = row.get('Soft%', 0)
        if pd.notna(soft_pct) and soft_pct < 1:
            soft_pct = soft_pct * 100
        elif pd.isna(soft_pct):
            soft_pct = 0

        ld_pct = row.get('LD%', 0)
        if pd.notna(ld_pct) and ld_pct < 1:
            ld_pct = ld_pct * 100
        elif pd.isna(ld_pct):
            ld_pct = 0

        gb_pct = row.get('GB%', 0)
        if pd.notna(gb_pct) and gb_pct < 1:
            gb_pct = gb_pct * 100
        elif pd.isna(gb_pct):
            gb_pct = 0

        fb_pct = row.get('FB%', 0)
        if pd.notna(fb_pct) and fb_pct < 1:
            fb_pct = fb_pct * 100
        elif pd.isna(fb_pct):
            fb_pct = 0

        # Apply park factors to base stats
        if team and pd.notna(team):
            pitcher_stats = {
                'ERA': era,
                'HR/FB': hr_fb * 100,  # Convert to percentage for adjustment
                'HBP%': hbp_pct
            }

            try:
                adjusted_stats = apply_park_factor_adjustments(
                    pitcher_stats, name, team, 'pitcher', year=2025
                )

                # Extract park-adjusted stats
                era = adjusted_stats.get('ERA', era)
                hr_fb = adjusted_stats.get('HR/FB', hr_fb * 100) / 100
                hbp_pct = adjusted_stats.get('HBP%', hbp_pct)

                if adjusted_stats.get('park_factor_adjustment', 1.0) != 1.0:
                    park_factors_applied += 1

            except Exception as e:
                # If park factors fail, use unadjusted stats
                pass

        # Calculate composites FRESH from (park-adjusted) base stats
        damage_control = calculate_damage_control_ratio(hr_fb, hard_pct, soft_pct, med_pct)
        contact_quality = calculate_contact_quality_index(hard_pct, med_pct, soft_pct, ld_pct, gb_pct, fb_pct)
        launch_quality = calculate_statcast_launch_quality(hr_fb, hard_pct)

        # Opportunity_Success fallback (keeping this from percentage features if available)
        opportunity_success = 0

        composites_calculated += 1

        # Build feature vector (10 features)
        features = [
            ip,
            bb_pct,
            k_pct,
            k_bb_pct,
            era,  # Park-adjusted
            damage_control,  # Fresh from park-adjusted stats with PERCENTAGE FIX
            opportunity_success,
            contact_quality,  # Fresh from park-adjusted stats with PERCENTAGE FIX
            hbp_pct,  # Park-adjusted
            launch_quality  # Fresh from park-adjusted stats with PERCENTAGE FIX
        ]

        feature_matrix.append(features)
        valid_players.append(row)
        player_names.append(name)

    print(f"  [PARK FACTORS] Applied to {park_factors_applied}/{len(pitcher_df)} pitchers")
    print(f"  [COMPOSITES] Calculated fresh for {composites_calculated}/{len(pitcher_df)} pitchers")
    print(f"  [PERCENTAGE FIX] Converted decimal batted ball stats to percentages")

    return {
        'valid_players': pd.DataFrame(valid_players),
        'feature_matrix': np.array(feature_matrix),
        'player_names': player_names
    }

try:
    first_half_pitchers = extract_pitcher_features_2025_fresh(first_half_pitchers_raw)

    if first_half_pitchers:
        print(f"\nProcessed First Half 2025 Pitchers:")
        print(f"  Valid players: {len(first_half_pitchers['valid_players'])}")
        print(f"  Feature matrix shape: {first_half_pitchers['feature_matrix'].shape}")
        print(f"  Features: 10 [IP, BB%, K%, K-BB%, ERA (park-adj), damage_control (fresh+fix), Opportunity_Success, Contact_Quality (fresh+fix), HBP% (park-adj), Launch_Quality (fresh+fix)]")
        print(f"  [FIX APPLIED] All 3 CSVs + fresh composites + park factors + percentage conversion")

except Exception as e:
    print(f"[ERROR] Error processing pitchers: {e}")
    first_half_pitchers = None
    import traceback
    traceback.print_exc()

print(f"\nFirst half 2025 data ready for Phase 1 with FRESH composites + FULL park factors + PERCENTAGE FIX!")
```
